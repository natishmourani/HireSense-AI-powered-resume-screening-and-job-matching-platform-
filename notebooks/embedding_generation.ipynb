# HireSense - Embedding Generation (Base Model)

This notebook loads the baseline embedding model (**all-MiniLM-L6-v2**) and generates embeddings for the resumes and job descriptions.

```python
import os
import json
import pickle
import numpy as np
from sentence_transformers import SentenceTransformer

# Load model
model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded!")
```

```python
# Load eval dataset
with open('../data/eval_dataset.json', 'r') as f:
    eval_pairs = json.load(f)

print(f"Loaded {len(eval_pairs)} evaluation pairs.")
```

```python
# Extract jobs and resumes text
jobs_text = [p['job'] for p in eval_pairs]
resumes_text = [p['resume'] for p in eval_pairs]

print("Generating embeddings (Base Model)...")
job_embeddings = model.encode(jobs_text, show_progress_bar=True)
resume_embeddings = model.encode(resumes_text, show_progress_bar=True)

# Save embeddings
os.makedirs('../data', exist_ok=True)
with open('../data/base_embeddings.pkl', 'wb') as f:
    pickle.dump({
        'job_embeddings': job_embeddings,
        'resume_embeddings': resume_embeddings
    }, f)
print("Base embeddings saved!")
```
