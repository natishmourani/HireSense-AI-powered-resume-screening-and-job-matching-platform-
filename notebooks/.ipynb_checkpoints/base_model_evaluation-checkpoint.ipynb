# HireSense - Base Model Evaluation

This notebook evaluates the baseline model's performance on the evaluation dataset using retrieval metrics.

```python
import json
import pickle
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Load eval dataset
with open('../data/eval_dataset.json', 'r') as f:
    eval_pairs = json.load(f)

# Load base embeddings
with open('../data/base_embeddings.pkl', 'rb') as f:
    embeddings = pickle.load(f)
    
job_embs = embeddings['job_embeddings']
resume_embs = embeddings['resume_embeddings']
```

```python
# Calculate cosine similarities
similarities = []
for i in range(len(eval_pairs)):
    sim = cosine_similarity([job_embs[i]], [resume_embs[i]])[0][0]
    similarities.append(float(sim))
    
labels = [p['label'] for p in eval_pairs]

# Calculate ranking, Precision, Recall, F1, Top-K
# We evaluate by treating each distinct job in the eval dataset as a query.
# In eval dataset, there are 10 jobs and 50 resumes (total 500 pairs).
# Let's reshape into query matrices.
num_jobs = 10
num_resumes = 50

sim_matrix = np.array(similarities).reshape(num_jobs, num_resumes)
label_matrix = np.array(labels).reshape(num_jobs, num_resumes)

precisions = []
recalls = []
f1s = []
mrr_list = []
top5_hits = 0

# Similarity threshold for classification (honest threshold)
threshold = 0.45

for i in range(num_jobs):
    sims = sim_matrix[i]
    lbls = label_matrix[i]
    
    # Sort indices by similarity descending
    sorted_indices = np.argsort(sims)[::-1]
    sorted_lbls = lbls[sorted_indices]
    
    # Classification metrics at threshold
    preds = (sims >= threshold).astype(int)
    tp = np.sum((preds == 1) & (lbls == 1))
    fp = np.sum((preds == 1) & (lbls == 0))
    fn = np.sum((preds == 0) & (lbls == 1))
    
    prec = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    rec = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * prec * rec / (prec + rec) if (prec + rec) > 0 else 0.0
    
    precisions.append(prec)
    recalls.append(rec)
    f1s.append(f1)
    
    # Top-K hit rate (Top-5)
    if np.sum(sorted_lbls[:5]) > 0:
        top5_hits += 1
        
    # MRR (Mean Reciprocal Rank)
    ranks = np.where(sorted_lbls == 1)[0]
    if len(ranks) > 0:
        mrr_list.append(1.0 / (ranks[0] + 1))
    else:
        mrr_list.append(0.0)

metrics = {
    "Precision": float(np.mean(precisions)),
    "Recall": float(np.mean(recalls)),
    "F1 Score": float(np.mean(f1s)),
    "Top-K Accuracy": float(top5_hits / num_jobs),
    "Mean Similarity Score": float(np.mean(sim_matrix[label_matrix == 1])),
    "Ranking Accuracy (MRR)": float(np.mean(mrr_list))
}

print("Base Model Metrics:")
print(json.dumps(metrics, indent=2))

with open('../data/base_metrics.json', 'w') as f:
    json.dump(metrics, f, indent=2)
```
